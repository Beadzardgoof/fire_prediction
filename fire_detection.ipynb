# Fire Detection and Risk Assessment

This notebook is designed to detect fires and assess fire risk using various data sources and machine learning techniques.


```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

# Geospatial libraries for shapefile reading
import geopandas as gpd
from shapely.geometry import Point

# Set up plotting style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
```

```python
# Load MODIS fire detection shapefile
# MODIS C6.1 provides global active fire detections from satellite imagery
shapefile_path = 'data/MODIS_C6_1_Global_24h.shp'

try:
    # Read the shapefile using geopandas
    fire_data = gpd.read_file(shapefile_path)
    
    print(f"Successfully loaded shapefile!")
    print(f"Number of fire detections: {len(fire_data)}")
    print(f"\nData columns:")
    print(fire_data.columns.tolist())
    print(f"\nFirst few rows:")
    print(fire_data.head())
    print(f"\nData info:")
    print(fire_data.info())
    print(f"\nCoordinate Reference System (CRS):")
    print(fire_data.crs)
    
except Exception as e:
    print(f"Error loading shapefile: {e}")
    print("Make sure geopandas is installed: conda install -c conda-forge geopandas")
```

## Data Loading and Exploration

In this section, we'll load and explore fire-related data. This could include:
- Weather data (temperature, humidity, wind speed)
- Historical fire incidents
- Geographic data
- Satellite imagery data


```python
# Explore the MODIS fire detection data structure
if 'fire_data' in locals():
    print("=== MODIS Fire Detection Data Summary ===\n")
    
    # Check for common MODIS attributes
    print("Available columns:")
    for col in fire_data.columns:
        print(f"  - {col}")
    
    # Display basic statistics for numeric columns
    print("\n=== Numeric Column Statistics ===")
    numeric_cols = fire_data.select_dtypes(include=[np.number]).columns
    if len(numeric_cols) > 0:
        print(fire_data[numeric_cols].describe())
    
    # Check geometry type
    print(f"\n=== Geometry Information ===")
    print(f"Geometry type: {fire_data.geometry.type.iloc[0] if len(fire_data) > 0 else 'N/A'}")
    print(f"Bounds (minx, miny, maxx, maxy): {fire_data.total_bounds}")
    
    # Sample of the actual data
    print(f"\n=== Sample Data (first 5 rows) ===")
    print(fire_data.head())
```

```python
# Visualize fire detections on a map
if 'fire_data' in locals() and len(fire_data) > 0:
    # Create a simple world map visualization
    fig, ax = plt.subplots(1, 1, figsize=(15, 10))
    
    # Plot fire detections
    fire_data.plot(ax=ax, markersize=1, color='red', alpha=0.6, label='Fire Detections')
    
    ax.set_title('MODIS Global Fire Detections (24h)', fontsize=16, fontweight='bold')
    ax.set_xlabel('Longitude', fontsize=12)
    ax.set_ylabel('Latitude', fontsize=12)
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print(f"Total fire detections plotted: {len(fire_data)}")
else:
    print("No fire data available to plot")
```
